```python
import numpy as np

oneDArray = np.array([1, 2, 3])

print(oneDArray)
print(type(oneDArray))
```

```python

twoDArray = np.array([[1, 2, 3], [4, 5, 6]])

print(twoDArray)
```

```python
threeDArray = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print(threeDArray)
```

```python
# dtype means datatype of the values/items inside the array
floatTypeArray = np.array([1, 2, 3], dtype=float) # This is the float type array

boolTypeArray = np.array([1, 2, 3], dtype=bool) # This is the boolean type array

print(floatTypeArray)
print(boolTypeArray)
```

```python
np.arange(1, 11) # Creates an ndarray (n dimensional array) starting from 1 to 10 as 11 is excluded [1, 11)
```

```python
np.arange(1, 11, 2) # Creates an ndarray from 1 to 10 but there is skip of 2
```

```python
withoutReshapeArray = np.arange(1, 11) # Shape is 1 x 10
print(withoutReshapeArray)

reshapedArray = withoutReshapeArray.reshape(5, 2) # Shape is 5 x 2
print(reshapedArray)
# It will only reshape if the reshaping is possible like the array which we want to reshape should be having the number of items = the reshape rows and columns
# For e.g we are having 10 values inside the withoutReshapeArray and we want to reshape to 5 x 2, so 10 = 5 * 2, so reshaping is possible
```

```python
np.arange(1, 11).reshape(5, 2)
```

```python
np.arange(1, 11).reshape(2, 5)
```

```python
np.ones((3, 4)) # If we want to create a n x m ndarray then we have to pass a tuple inside the function
```

```python
np.zeros(3) # If we want to create a vector or 1 x n zeros ndarray of items 0 so you should only give the columns size like here we have given 3 so it is 1 x 3 ndarray
```

```python
np.random.random((3, 4)) # It will generate a ndarray of items from 0 to 1 of the size you have passed in the tuple
# The range of the random will be from [0, 1) -> 1 excluded, 0 included
```

```python
np.linspace(-10, 10, 10) # It will generate a ndarray of size 10 (the third parameter) from range starting from -10 to ending point 10
# It will generate 10 number of items in a ndarray and each point will be of same distance
```

```python
np.linspace(-1, 10, 20)
# The full form of linspace is linear space
# Every point distance will be same in the linspace
```

```python
np.identity(3)
# It will create an identity matrix of size 3 x 3
```
