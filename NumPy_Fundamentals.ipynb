# Array Attributes

```python
a1 = np.arange(10)
a2 = np.arange(12, dtype=float).reshape(3, 4)
a3 = np.arange(8).reshape(2, 2, 2)
```

```python
# ndim tells about the dimension of the ndarray, it's fullform is Number Of Dimensions
print(a1.ndim) # 1D Array -> Vector
print(a2.ndim) # 2D Array -> Matrix
print(a3.ndim) # 3D Array -> Tensor
```

```python
# shape tells us about the shape of the given ndarray
print(a1.shape) # (10,) as it has 10 columns no rows
print(a2.shape) # (3, 4) as it has 3 rows 4 columns
print(a3.shape) # (2, 2, 2) as it has 2 2D arrays each having 2 rows and 2 columns
```

```python
# size tells us how many items are there in a ndarray
print(a1.size)
print(a2.size)
print(a3.size)
```

```python
# itemsize tells us about the size of one array item in bytes like how much bytes of size is taken by one item of the ndarray
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
# Here all 3 ndarrays are giving 8 itemsize as the default int is int64 is of 8 bytes of size
```

```python
# dtype will give you the datatype of items in the ndarray
print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

# Array Operations

```python
a1 = np.arange(12).reshape(3, 4)
a2 = np.arange(12, 24).reshape(3, 4)
```

```python
a1 * 2
```

```python
# Will check this operation for each item in the ndarray a2 and then tell whether for each item in the array that the operation or check is true or not
a2 > 18
```

```python
a2 == 18
```

```python
# Vector operations as the size of a1 and a2 are same hence we can do the vector operations for ndarray
a1 + a2
```

```python
a1, a2
```

```python
# It will raise the power of corresponding item of the a2 to the a1
a1 ** a2
```

# Array Functions

```python
a1 = np.random.random((3, 3))
a1 = np.round(a1 * 100)
a1
```

```python
np.max(a1)
```

```python
np.min(a1)
```

```python
# Will give the sum of all the items in a1 ndarray
np.sum(a1)
```

```python
# Will give the product of all the items in a1 ndarray
np.prod(a1)
```

```python
# axis = 0 means that we are doing it for column, so for each column it gives us the min item in a1 ndarray
np.min(a1, axis=0)
```

```python
# axis = 1 means that we are doing it for row, so for each row it gives us the min item in a1 ndarray
np.min(a1, axis=1)
```

```python
# axis = 1 means row
# axis = 0 means column
```

```python
# It will calculate the mean for all items in the ndarray
np.mean(a1)
```

```python
np.mean(a1) == np.sum(a1) / a1.size
```

```python
# It will give us the mean of a1 matrix column wise
np.mean(a1, axis=0)
```

```python
np.mean(a1, axis=0) == np.sum(a1, axis=0) / np.size(a1, axis=0)
```

```python
np.size(a1, axis=0)
```

```python
# Median
np.median(a1)
```

```python
# Standard deviation
np.std(a1, axis=0)
```

```python
np.var(a1)
```

```python
# Whenever we are taking dot product of two matrices so the columns of the first matrix should be equal to the rows of the second matrix
a1 = np.arange(12).reshape(3, 4)
a2 = np.arange(12, 24).reshape(4, 3)
# Here the a1 matrix has 3 x 4 and a2 matrix has 4 x 3 so the resultant matrix has shape of 3 x 3
np.dot(a1, a2)
```

```python
a1 = np.random.random((3, 4)) * 100
a1
```

```python
# log is used to take log of each item in the ndarray
np.log(a1)
```

```python
# exp is used to calculate the exponent of each item in the ndarray
np.exp(a1)
```

```python
# round takes the round for each item in the ndarray
np.round(np.random.random((2, 3)) * 100)
```

```python
# floor takes the floor for each item in the ndarray
# e.g 3.4 -> 3 it takes the lower value of the float
np.floor(np.random.random((2, 3)) * 100)
```

```python
# ceil takes the ceil for each item in the ndarray
# e.g 3.4 -> 4 it takes the higher value of the float
np.ceil(np.random.random((2, 3)) * 100)
```
