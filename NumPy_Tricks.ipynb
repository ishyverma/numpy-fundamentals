```python
import numpy as np
```

```python
# np.sort
# return a sorted copy of an array
```

```python
a = np.random.randint(1, 100, 15)
a
```

```python
b = np.random.randint(1, 100, 24).reshape(6, 4)
b
```

```python
# Gives the sorted array in ascending order
np.sort(a)
```

```python
# It will sort the 2D array in the row wise manner
np.sort(b)
```

```python
# It will sort the 2D array in the column wise manner
np.sort(b, axis=0)
```

```python
# To sort the array in the descending order we have to use this [::-1]
np.sort(a)[::-1]
```
