```python
import numpy as np
```

```python
import time
a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]
c = []
start = time.time()
for i in range(len(a)):
  c.append(a[i] + b[i])
print(time.time() - start)
```

```python
a = np.arange(10000000)
b = np.arange(10000000, 20000000)
start = time.time()
c = a + b
print(time.time() - start)
```

```python
# It tells that numpy ndarrays are faster than python lists
```

# Broadcasting

```python
a = np.arange(6).reshape(2, 3)
b = np.arange(6, 12).reshape(2, 3)

print(a + b)

print(a)
print(b)
```

```python
a = np.arange(6).reshape(2, 3)
b = np.arange(3).reshape(1, 3)

print(a + b)

print(a)
print(b)
```

# Working with mathematical formulas

```python
a = np.arange(10)
np.sin(a)
```

```python
def sigmoid(array):
  return 1 / (1 + np.exp(-array))

sigmoid(a)
```

```python
def mse(actual, predicted):
  return np.mean((actual - predicted) ** 2)

actual = np.random.randint(1, 50, 25)
predicted = np.random.randint(1, 50, 25)

mse(actual, predicted)
```

```python
def bce(actual, predicted_probability):
  log_p = np.log(predicted_probability)
  one_log_p = np.log(1 - predicted_probability)
  return -np.mean((actual * log_p) + ((1 - actual) * one_log_p))

actual = np.random.randint(0, 2, 25)
predicted = np.random.random(25)
bce(actual, predicted)
```

# Working with missing values
