```python
import numpy as np
```

```python
import time
a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]
c = []
start = time.time()
for i in range(len(a)):
  c.append(a[i] + b[i])
print(time.time() - start)
```

```python
a = np.arange(10000000)
b = np.arange(10000000, 20000000)
start = time.time()
c = a + b
print(time.time() - start)
```

```python
# It tells that numpy ndarrays are faster than python lists
```

```python
import sys
a = [i for i in range(10000000)]
sys.getsizeof(a)
```

```python
a = np.arange(10000000, dtype=np.int8)
sys.getsizeof(a)
```

# Advanced Indexing

```python
a = np.arange(12).reshape(4, 3)
a
```

```python
a[::2]
```

```python
# Fancy indexing: We can directly put the indexes of the row we want and we will get them
a[[0, 2, 3]]
```

```python
a = np.arange(24).reshape(6, 4)
a
```

```python
a[[0, 2, 3, 5]]
```

```python
a[:, [0, 2, 3]]
```

```python
# Fancy indexing is used whenever we are unable to find pattern in indexing
```

```python
a = np.random.randint(1, 100, 24).reshape(6, 4)
a
```

```python
a > 50
```

```python
# This is boolean masking
a[a > 50]
```

```python
a[a % 2 == 0]
```

```python
# When working with booleans we should use & as it do the bitwise operations
a[(a > 50) & (a % 2 == 0)]
```

```python
a[a % 7 == 0]
```
